```sql
%python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```sql
%python
dbutils.widgets.removeAll()
```

```sql
%python
dbutils.widgets.text("1_storageName", "adlsmartdata0912")
dbutils.widgets.text("2_container", "bronze")
dbutils.widgets.text("3_catalog", "catalog_smartdata")
dbutils.widgets.text("4_schema", "bronze")
```

```sql
%python
storage_name = dbutils.widgets.get("1_storageName")
container = dbutils.widgets.get("2_container")
catalog =  dbutils.widgets.get("3_catalog")
schema =  dbutils.widgets.get("4_schema")
```

```sql
%python

path_base = f"abfss://{container}@{storage_name}.dfs.core.windows.net/"
path_cliente =  f"{path_base}Cliente.csv"
```

```sql
%python
cliente_schema = StructType(fields=[
    StructField("ClienteId", IntegerType(), False),
    StructField("Nombre", StringType(), False),
    StructField("Provincia", StringType(), False),
    StructField("EstadoCivil", StringType(), False),
    StructField("FechaNacimiento", StringType(), False),
    StructField("IngresoAnual",IntegerType(), False)
    
])
```

```sql
%python
df_cliente =  spark.read.option('header', True)\
                        .schema(cliente_schema)\
                        .csv(path_cliente)
     
```

```sql
%python

df_cliente.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.cliente")
```
