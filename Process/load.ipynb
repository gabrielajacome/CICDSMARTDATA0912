```sql
%python
dbutils.widgets.text("1_storageName", "adlsmartdata0912")
dbutils.widgets.text("2_container", "bronze")
dbutils.widgets.text("3_catalog", "catalog_smartdata")
dbutils.widgets.text("4_schema", "silver")
dbutils.widgets.text("5_schema", "golden")
```

```sql
%python
storage_name = dbutils.widgets.get("1_storageName")
container = dbutils.widgets.get("2_container")
catalog =  dbutils.widgets.get("3_catalog")
schema_origen =  dbutils.widgets.get("4_schema")
schema_destino =  dbutils.widgets.get("5_schema")
```

```sql
%python
df_cliente_producto =  spark.table(f"{catalog}.{schema_origen}.cliente_producto")
df_kpi_cliente_propas_ingreso =  spark.table(f"{catalog}.{schema_origen}.kpi_cliente_propas_ingreso")
df_kpi_productoactivo_estado =  spark.table(f"{catalog}.{schema_origen}.kpi_productoactivo_estado")
df_kpi_productopasivo_estado   =  spark.table(f"{catalog}.{schema_origen}.kpi_productopasivo_estado")
```

```sql
%python
df_cliente_producto.write.mode("overwrite").saveAsTable(f"{catalog}.{schema_destino}.cliente_producto")
df_kpi_cliente_propas_ingreso.write.mode("overwrite").saveAsTable(f"{catalog}.{schema_destino}.kpi_cliente_propas_ingreso")
df_kpi_productoactivo_estado.write.mode("overwrite").saveAsTable(f"{catalog}.{schema_destino}.kpi_productoactivo_estado")
df_kpi_productopasivo_estado.write.mode("overwrite").saveAsTable(f"{catalog}.{schema_destino}.kpi_productopasivo_estado")
```
