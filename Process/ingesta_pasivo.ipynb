```sql
%python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```sql
%python
dbutils.widgets.removeAll()
```

```sql
%python
dbutils.widgets.text("1_storageName", "adlsmartdata0912")
dbutils.widgets.text("2_container", "bronze")
dbutils.widgets.text("3_catalog", "catalog_smartdata")
dbutils.widgets.text("4_schema", "bronze")
```

```sql
%python
storage_name = dbutils.widgets.get("1_storageName")
container = dbutils.widgets.get("2_container")
catalog =  dbutils.widgets.get("3_catalog")
schema =  dbutils.widgets.get("4_schema")
```

```sql
%python

path_base = f"abfss://{container}@{storage_name}.dfs.core.windows.net/"
path_pasivo =  f"{path_base}Pasivo.csv"
```

```sql
%python
pasivo_schema = StructType(fields=[
    StructField("ClienteId", IntegerType(), False),
    StructField("PasivoId", StringType(), False),
    StructField("SaldoPasivo", IntegerType(), False)
    
])
```

```sql
%python
df_pasivo =  spark.read.option('header', True)\
                        .schema(pasivo_schema)\
                        .csv(path_pasivo)
```

```sql
%python
df_pasivo.write.mode("overwrite").saveAsTable(f"{catalog}.{schema}.pasivo")
```
